```python
# import numpy, pandas and seaborn
import numpy as np
import pandas as pd

# install matplotlib and import
!pip install matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# read file
mushroom = pd.read_csv("mushrooms.csv")
```

```python
# Compare multiple characteristics against class
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Cap shape vs class
pd.crosstab(mushroom['cap-shape'], mushroom['class']).plot(kind='bar', ax=axes[0,0])
axes[0,0].set_title('Cap Shape vs Class')
axes[0,0].set_xlabel('Cap Shape')

# Cap color vs class
pd.crosstab(mushroom['cap-color'], mushroom['class']).plot(kind='bar', ax=axes[0,1])
axes[0,1].set_title('Cap Color vs Class')
axes[0,1].set_xlabel('Cap Color')

# Odor vs class
pd.crosstab(mushroom['odor'], mushroom['class']).plot(kind='bar', ax=axes[1,0])
axes[1,0].set_title('Odor vs Class')
axes[1,0].set_xlabel('Odor')

# Habitat vs class
pd.crosstab(mushroom['habitat'], mushroom['class']).plot(kind='bar', ax=axes[1,1])
axes[1,1].set_title('Habitat vs Class')
axes[1,1].set_xlabel('Habitat')

plt.tight_layout()
plt.show()
```

```python
#piechart comparing class of population
class_counts = mushroom['class'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(class_counts, labels=class_counts.index, autopct='%1.1f%%', startangle=90, colors=['lightblue','lightcoral'])
plt.title("Class Proportion of Population")
plt.show()
```

```python
# Create a crosstab/contingency table
ct = pd.crosstab(mushroom['habitat'], mushroom['class'])
sns.heatmap(ct, annot=True, fmt='d', cmap='Reds')
```
